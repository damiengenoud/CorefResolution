Train the model. Default save folder: saved_models/my_models (can be changed in config.py). To use an already existing mentions extractor, you can pass as argument "require_extractor_training=False" and "model_save_folder_path=path", with path being the path to the folder containing a trained mentions_extractor.pt. Do not set "require_extractor_training=True" while specifying "model_save_folder_path=path", as this would train an extractor that would not be used by the model.

Evaluate the model. By default, evaluates the pre-trained models (in saved_models/pretrained_models), but you can pass as argument to evaluate() a path to folder that contains a trained coref_model.pt to evaluate it.
